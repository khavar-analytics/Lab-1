### Task 4

```python
user_events = [
    {"user_id": 1, "event_type": "login", "duration_seconds": 5},
    {"user_id": 1, "event_type": "view_page", "duration_seconds": 120},
    {"user_id": 2, "event_type": "login", "duration_seconds": "ab"},
    {"user_id": 2, "event_type": "view_page", "duration_seconds": 95},
    {"user_id": 2, "event_type": "logout", "duration_seconds": 3},
    {"user_id": 3, "event_type": "login", "duration_seconds": 4},
    {"user_id": 3, "event_type": "purchase", "duration_seconds": 240},
    {"user_id": 1, "event_type": "purchase", "duration_seconds": 180},
    {"user_id": 4, "event_type": "login", "duration_seconds": -5},
    {"user_id": 4, "event_type": "view_page", "duration_seconds": 60},
    {"user_id": 4, "event_type": "view_page", "duration_seconds": 75},
    {"user_id": 3, "event_type": "logout", "duration_seconds": 2}
]
```

```python
cleaned_events = []

for event in user_events:
    try:
        duration = float(event["duration_seconds"])
        if duration >= 0:
            cleaned_event = event.copy()
            cleaned_event["duration_minutes"] = duration / 60
            cleaned_events.append(cleaned_event)
    except (ValueError, TypeError):
        continue

            
```

```python
cleaned_events
```

### Task 5

#### event counts

```python
event_counts = {}

for i in cleaned_events:
    event_type = i["event_type"]
    event_counts[event_type] = event_counts.get(event_type, 0) + 1
print(event_counts)
```

#### average druration minutes

```python
duration_sum = {}
event_count = {}

for x in cleaned_events:
    event_type = x["event_type"]
    duration = x["duration_minutes"]

    duration_sum[event_type] = duration_sum.get(event_type, 0) + duration
    event_count[event_type] = event_count.get(event_type, 0) + 1
```

```python
avg_duration_minutes = {}

for event_type in duration_sum:
    avg_duration_minutes[event_type] = (duration_sum[event_type] / event_count[event_type])

print(avg_duration_minutes)
```

#### unique users

```python
unique_users = set()

for event in cleaned_events:
    unique_users.add(event["user_id"])

unique_user_count = len(unique_users)

print(unique_user_count)
```

#### CSV-formatted string

```python
lines = []
lines.append("metric,key,value")

# event counts
for event_type, count in event_counts.items():
    lines.append(f"event_count,{event_type},{count}") 

# average durations
for event_type, avg in avg_duration_minutes.items():
    lines.append(f"avg_duration_minutes,{event_type},{avg}") 

# unique users
lines.append(f"unique_users,all,{unique_user_count}")  

csv_string = "\n".join(lines)
print(csv_string)
```
